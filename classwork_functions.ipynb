```python
def add(a,b):
    return a + b
print(add(2,3))
```

```python
def add(*args):
    print(args)
add(1,2,3,4,5)
def add(*args):
    print(sum(args))
add(1,2,3,4,5)
```

```python
def student_info(**kwargs):
    return kwargs
print(student_info(name="John", age=25, city="New York"))
```
