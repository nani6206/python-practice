```python
import numpy as np
a=np.array([1, 2, 3])
print(a)
b=np.array([[4, 5, 6],[7,8,9]])
print(b)
c=np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(c)
print(f'{a.ndim} {b.ndim} {c.ndim}')
print(f'shape:{c.shape} size:{c.size} dtype:{c.dtype} itemsize:{c.itemsize} nbytes:{c.nbytes}')
```

```python
print(f"zeros: {np.zeros((2,3))}")
print(f"ones: {np.ones((2,3))}")
print(f"full: {np.full((2,3),7)}")
print(f"Arrange: {np.arange(0,10,2)}")
print(f"Linspace: {np.linspace(0,1,5)}")
```

```python
a=np.array([1,2,3,4,5],dtype=np.int32)
b=np.array([1.5,2.5,3.5],dtype=np.float32)
c=np.array([True,False,True])
print(f"a dtype: {a.dtype} b dtype: {b.dtype} c dtype: {c.dtype}")
x=a.astype(np.float64)
print(x,x.dtype)
print(f"first element of a:{a[0]} using idexing")
```

```python
matrix=np.array([[1,2,3],
                 [4,5,6]])
print(f"first element of matrix:{matrix[0,0]} using indexing row num and column num")
```

```python
#for 1-D
print(a[:1])
print(a[-1])
print(a[::-1])
# for 2D
print(matrix[1,:])
```

```python
arr=np.arange(1,13)
print(arr)
m=arr.reshape(3,4)
print(m)
cube=arr.reshape(2,2,3)
print(cube)
print(cube.shape)
at=arr.reshape(3,-1)
print(at)
```

```python
#copy-immutable[flatten2dto1d] view-mutable[ravel-2dto1d]
n=np.arange(1,6)
v=n[:]
print(n)
v[0]=10
print(n)
t=np.arange(1,6)
print(t)
copy=t.copy()
copy[0]=10
print(t)
```

```python
# flatten changes not happen
print(matrix)
flat=matrix.flatten()
flat[0]=90
print(matrix)
#ravel 2d to 1d affected
print(matrix)
rav=matrix.ravel()
rav[0]=90
print(matrix)
```

```python
print(matrix)
print(f"Trandeposed {matrix.T} \nshape:{matrix.T.shape}")
```

```python
a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
print(f"vectorised{a+b}")
print(f"vectorised{a-b}")
print(f"vectorised{a*b}")
#scaler oper
print(a)
print(a+5)
print(a-5)
print(a*5)
```

```python
print(f"sqrt:{np.sqrt(a)}")
print(f"exp:{np.exp(a)}")
print(f"abs:{np.abs(np.array([-2,-3,2]))}")
print(f"log:{np.log(np.array([1,np.e,np.e**2]))}")
print(f"floor:{np.floor(np.array([1.2,3.4]))}")
print(f"floor:{np.ceil(np.array([1.2,3.4]))}")
```

```python
scores = np.array([10,20,30,40,50])
print(f"sum:{np.sum(scores)}")
print(f"mean:{np.mean(scores)}")
print(f"median:{np.median(scores)}")
print(f"max:{np.max(scores)}")
print(f"sum:{np.min(scores)}")
```

```python
classa=[48,49,50,51,52]
classb=[20,35,50,65,80]
print(f"mean:{np.mean(classa)}")
print(f"mean:{np.mean(classb)}")
print(f"variance:{np.var(classa)}")
print(f"var:{np.var(classb)}")
print(f"stddev:{np.std(classa)}")
print(f"stddev:{np.std(classb)}")
```
