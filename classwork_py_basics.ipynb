```python
# string methods
# Common String Functions in Python

text = "Python Programming"

print("Original String:", text)

# Convert all letters to uppercase
print("\nupper():", text.upper())

# Convert all letters to lowercase
print("lower():", text.lower())

# Convert the first letter of every word to uppercase
print("title():", text.title())

# Convert only the first letter to uppercase
print("capitalize():", text.capitalize())

# Swap uppercase and lowercase letters
print("swapcase():", text.swapcase())

# Find the length of the string
print("len():", len(text))

# Replace a word
print("replace():", text.replace("Python", "Java"))

# Split the string into a list
print("split():", text.split())

# Join list elements into a string
subjects = ["AI", "ML", "Python"]
print("join():", " ".join(subjects))

# Remove spaces from both sides
name = "   pythin Programming    "
print("strip():", name.strip())

# Remove spaces from the left
print("lstrip():", name.lstrip())

# Remove spaces from the right
print("rstrip():", name.rstrip())

# Find the position of a word
print("find():", text.find("Program"))

# Find the position (raises an error if not found)
print("index():", text.index("Program"))

# Count occurrences of a character
print("count():", text.count("m"))

# Check if the string starts with a value
print("startswith():", text.startswith("Python"))

# Check if the string ends with a value
print("endswith():", text.endswith("ing"))

# Check whether all characters are alphabets
print("isalpha():", "Python".isalpha())

# Check whether all characters are digits
print("isdigit():", "12345".isdigit())

# Check whether all characters are letters or numbers
print("isalnum():", "Python123".isalnum())

# Check whether the string contains only spaces
print("isspace():", "     ".isspace())

# Add leading zeros
print("zfill():", "25".zfill(5))

# Center the string
print("center():", text.center(30, "-"))

# Left align the string
print("ljust():", text.ljust(30, "."))

# Right align the string
print("rjust():", text.rjust(30, "."))

# Split into three parts using the separator
print("partition():", text.partition(" "))

# Split from the right side
print("rpartition():", text.rpartition(" "))

# Convert the string into bytes
print("encode():", text.encode())

# Convert to lowercase (more aggressive than lower)
print("casefold():", "PYTHON".casefold())

# Format a string
print("format():", "Welcome {} to {}.".format("Mariya", "Python"))
```
